# Member 1: data pipeline
Builds candidate segments + features for train / student_test (never evaluator data), then checks quality. Run from the repo root.

```
import subprocess, sys
subprocess.run([sys.executable, "-m", "src.member1_pipeline", "--data-root", ".", "--output", "results/member1/data"], check=True)
```

```
import pandas as pd, matplotlib.pyplot as plt
inv = pd.read_csv("results/member1/data/inventory.csv")
inv.groupby("split")[["n_observations","n_candidates","no_candidates","gaps_over_30s","unreliable_alignment"]].sum()
```

```
print(inv.groupby("split").candidate_miss_rate.describe())
ax = inv.boxplot(column="candidate_miss_rate", by="split"); plt.title("Candidate miss rate per trip"); plt.suptitle("")
plt.savefig("results/member1/candidate_miss_rate.png", dpi=130); plt.show()
```

```
cand = pd.read_parquet("results/member1/data/candidates.parquet")
print(cand.shape, cand.label.value_counts().to_dict())
print(cand.isna().sum()[lambda s: s>0])
print(cand.groupby("split").label.mean())
```
